```python
# --- exemples d'utilisation de modules de la bibliothèque standard ------
# exemple de datetime
# importer l'objet datetime à partir du module datetime

from datetime import datetime, timedelta, time

# créer une variable dt représente la date d'aujourd'hui avec la signature de base 

dt = datetime(2026, 11, 10, 15, 30)
# idem avec une fonction interne

now = datetime.now()

# idem à partir de la chaine de caractère "2026-04-01 15:30" et le format "%Y-%m-%d %H:%M" 
dt = datetime.strptime("2026-11-10 15:30", "%Y-%m-%d %H:%M")

# à partir de la variable, afficher l'année, la date au format "%d/%m/%Y", le nb de secondes à partir le 1er janvier 1970
dt.year, dt.strftime("%d/%m/%Y"), dt.timestamp()

# afficher la durée entre la fin de la journée et maintenant, en heures, min, s
# pour faire aujourd'hui à 17h30 qqsoit le jour
# 1/ créer la date aujourd'hui à partir de now
# 2/ un horaire en dehors d'une date ==> from datetime import time != from time import time
# 3/ combine la date et l'horaire
today_date = datetime.date(datetime.now()) # datetime.today() EST UNE FRAUDE !!!
time_1730 = time(17, 30)
today_1730 = datetime.combine(today_date, time_1730)

duration = today_1730 - datetime.now()
hours, mins, secs =  str(duration).split(":")
print(f"il reste {hours}h {mins}m et {int(float(secs))}s")
# date + duree (timedelta) = date
```
