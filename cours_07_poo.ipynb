```python
# un exemple en programmation impérative / procédurale 
# sans programmation orientée objet

## initialisations

# ce dictionnaire représente la strucutre générique d'un compte
account = {
  "balance": 0,
  "overdraft": 0,
}

# créer une fonction init_account qui va mettre à jour les champs d'un dict account en param
# à partir de paramètres balance et overdraft
def init_account(account: dict, balance: float, overdraft: float) -> dict:
  account["balance"] = balance
  account["overdraft"] = overdraft
  return account

# créer une fonction withdraw pour effectuer un retrait d'un montant sur un compte en param.
def withdraw(account: dict, amount: float):
  if amount < 0:
    print(f"Transaction refusée: {amount} négatif")
  elif amount > account["balance"] + account["overdraft"]:
    print(f"Transaction refusée: {amount} fonds insuffisants")
  else:
    account["balance"] -= amount
    print(f"Transaction acceptée")

## programme principal
if __name__ == "__main__":
  # initialiser le dictionnaire account avec un solde de 1000 et un découvert de 200
  personal_account = init_account(account, balance=1000, overdraft=200)
  # saisir un montant, effectuer le retrait et afficher le solde après
  amount = input("Entrez un montant : ")
  amount = int(amount)

  withdraw(personal_account, amount)
  print(f"nouveau solde: {personal_account["balance"]}")
```

```python
# --- même exemple en Programmation Orientée Objet (POO) ---------
class Account:
    balance: float = 0.
    overdraft: float = 200
 
    def withdraw(self, amount: float):
        if amount < 0:
            print(f"Transaction refusée: {amount} négatif")
        elif amount > self.balance + self.overdraft:
          print(f"Transaction refusée: {amount} fonds insuffisants")
        else:
          self.balance -= amount
          print(f"Transaction acceptée")
## programme principal
if __name__ == "__main__":
   acc = Account()
   print(acc.balance, acc.overdraft)
   # Preuve 1: muable !! Preuve 2: et les attributs sont publics par défaut !!
   acc.balance = 1000
   
   # preuve 3: les attributs d'objets/d'instance sont différents des attributs de classe 
   print(acc.balance, Account.balance)

   # on ne demande pas self à l'appel car self est l'objet instancié lui même (acc)
   acc.withdraw(500)
   print(acc.balance)
```

```python
# --------------------- utilisation de la méthode magique __init__() ------------

class Account:
    """
    classe de base de gesitonde compte dépôt
    """
    # un attribut/méthode de forme __xxxx__ est un attribut/méthode MAGIQUE 
    # en utilisant __init__ on a pas forcément besoin d'attributs de classe
    def __init__(self, balance: float, overdraft:float = 200):
       self.balance  = balance
       self.overdraft = overdraft
 
    def withdraw(self, amount: float):
        if amount < 0:
            print(f"Transaction refusée: {amount} négatif")
        elif amount > self.balance + self.overdraft:
          print(f"Transaction refusée: {amount} fonds insuffisants")
        else:
          self.balance -= amount
          print(f"Transaction acceptée")


## programme principal
if __name__ == "__main__":
  # __init__ permet créer/d'alimenter les attributs d'objets à l'instanciation
  acc = Account(1000)
  print(acc.balance, acc.overdraft)
  acc.withdraw(500)
  print(acc.balance)

```

```python
# ------------------------ exemple client ---------------------------------

from datetime import datetime
# créer une classe client qui contient firstname, name, et date_joint
# et 2 méthodes: 
#     get_full_name: retourne le prénom capitalisé et le nom en majuscule
#     get_date_joint: retourne la date dans le format voulu en paramètre

class Client:
    def __init__(self, firstname: str, name: str, date_joint: datetime):
        self.firstname = firstname
        self.name = name
        self.date_joint = date_joint

    def get_full_name(self) -> str:
        return f"{self.firstname.capitalize()} {self.name.upper()}"

    def get_date_joint(self, format: str = "%d/%m/%Y") -> str:
        return self.date_joint.strftime(format)

    def __str__(self) -> str:
        return f"{self.get_full_name()}, date_joint={self.get_date_joint()})"
```

```python
# -------------------------- héritage simple -----------------------------

class Person:
    def __init__(self, firstname: str, name: str):
        self.firstname = firstname
        self.name = name

    def get_full_name(self) -> str:
        return f"{self.firstname.capitalize()} {self.name.upper()}"

    def __str__(self):
        return self.get_full_name()
# reprendre l'exemple précédent sachant qu'un client EST une personne avec un prénom et un nom
## person est considérée comme classe parente de client
## client est //                      enfant de personne , hérite de personne

class Client(Person):

    def __init__(self, firstname: str, name: str, date_joint: datetime):
        ## super().<methode>() permet d'exécuter une méthode de la classe PARENT 
        ## dans la classe ENFANT et sur l'objet ENFANT
        super().__init__(firstname, name)
        self.date_joint = date_joint        
    
    def get_date_joint(self, format: str = "%Y-%m-%d") -> str:
        return self.date_joint.strftime(format)

    def __str__(self):
        return super().__str__() + " date_joint: " + self.get_date_joint() 



## la classe enfant peut réutiliser directement les méthodes parentes
## la classe enfant peut également créer ses propres méthodes


# classe personne
#    utilise prénom et nom

# classe client est une personne
#    utilise prénom, nom et date_joint      
```

```python
# ------------------------- injection de dépendances -----------------------------
from datetime import datetime

class Account:
    def __init__(
            self, 
            balance: float, 
            client: Client, 
            overdraft: float = 200
    ):
        self.balance = balance
        self.client = client
        self.overdraft = overdraft

    def bonus_10years(self) -> None:
        """
        ajoute 100 si c'est l'anniv des 10 ans
        """
        now = datetime.now()
        if (
            self.client.date_joint.month == now.month 
            and self.client.date_joint.day == now.day
            and now.year - self.client.date_joint.year == 10
        ) :
            self.balance += 100

cl = Client("matt", "lamam", datetime.strptime("2016-11-12", "%Y-%m-%d"))
acc = Account(1000, cl)

acc.bonus_10years()
acc.balance

# relation de type AVOIR entre une classe "utilisateur" et une autre qui est une dépendance
# l'utilisateur mange sa dépendance à l'instanciation et utilise LA SIGNATURE PUBLIQUE de cette dépendance
```

```python
# ------------------------- héritage multiple + polymorphisme --------------------

from abc import ABC, abstractmethod

class IAccount(ABC):
    @abstractmethod
    def get_balance(self) -> float: pass

class BasicAccount(IAccount):
    def __init__(self, balance: float):
        self.balance = balance

    def get_balance(self):
        return self.balance

    def withdraw(self, amount: float):
        if amount < 0:
            print(f"Transaction refusée: {amount} négatif")
        elif amount > self.balance:
          print(f"Transaction refusée: {amount} fonds insuffisants")
        else:
          self.balance -= amount
          print(f"Transaction acceptée")

# acc = BasicAccount(1000)
# print(acc.get_balance())

class SavingAccount(IAccount):
    def __init__(self, balance: float, interest_rate: float=1):
        self.balance = balance
        self.interest_rate = interest_rate

    def get_balance(self):
        return self.balance

    def apply_interest(self):
        self.balance += self.balance * self.interest_rate / 100

class PremiumAccount(BasicAccount, SavingAccount):
    """
    héritage multiple
    
    pour accéder à une méthode parente
    1/ super qui respecte l'algorithme MRO (Method Resolution Order)
        règle 1 ==> première classe parente
    2/ exécuter une méthode depuis une des classes mais ne pas oublier de demander self en 1er dans l'appel !!
    """
    def __init__(self, balance, interest_rate = 1):
       # super().__init__(balance, interest_rate) # si Saving en 1er !!!
       SavingAccount.__init__(self, balance, interest_rate)


acc  = PremiumAccount(1000)

acc.withdraw(200)
acc.get_balance()
acc.apply_interest()
acc.get_balance()
        
```

```python
# ------------------------ méthodes magiques ------------------------
from typing import List
from functools import singledispatchmethod

class Account:
    def __init__(self, b, num: int):
        self.balance = b
        self.num = num

    def __str__(self):
        return f"num: {self.num}, balance: {self.balance}"

class Client:
    def __init__(self, n: str, accounts: List[Account]):
        self.name = n
        self.accounts = accounts

    def __str__(self):
        accounts = ""
        for acc in self.accounts:
            accounts += str(acc) + ','
        
        return f"name: {self.name}: accounts: {accounts[:-1]}"

    @singledispatchmethod
    def __getitem__(self, num):
        # Ce bloc s'exécute si le type n'est ni int ni slice
        raise TypeError(f"Type d'index non supporté : {type(num).__name__}")
    
    @__getitem__.register
    def _(self, num: int) -> Account:
        """
        pilote le []
        """
        return list(filter(lambda acc: acc.num == num, self.accounts))[0]

    @__getitem__.register
    def _(self, num: slice) -> List[Account]:
        return self.accounts[num]

acc = Account(1000, 1)
acc2 = Account(2000, 2)
cl = Client("matt", [acc, acc2])
print(cl)
print(cl[2].balance)
print(cl[1:])
# print(cl["machin"])
```

```python
# -------------- itérateur / itérable ---------------

class MyRange:
    def __init__(self, limit: int=10):
        """
        condition d'arrêt
        """
        self.limit = limit

    def __iter__(self):
        """
        condition d'(ré)initialisation
        """
        self.counter = 0
        return self

    def __next__(self):
        """
        condition de chgt
        """
        if self.counter < self.limit:
            ret = self.counter
            self.counter += 1
            return ret
        else:
            raise StopIteration

for i in MyRange():
    print(i)

# r est un itérable !!!
r = MyRange(3)

# r est un itérateur avec compteur (ré)initialisé
it = iter(r)

next(it)
next(it)
next(it)
# next(it)

#
```
