```python
# ---------- définition et appel d'une fonction -----------------------
### une fonction est un objet créé/définie avec: 
### le mot clé def suivi de <nom_de_la_fonction>(): 
###     et après un bloc d'instructions  

# définition d'une fonction ma_fonction() qui contient l'affichage de "coucou"
def ma_fonction():
    print("coucou")

### pour exécuter le bloc de la définition, on doit appeler la fonction 
### <nom_de_la_fonction>()
### (): opérateur d'appel
### une fonction est un type de donnée dite "callable"

# afficher l'appel de la fonction ma_fonction
print("appel de ma_fonction() => ", ma_fonction())
```

```python
# ----------- retour d'une fonction ------------------------------

# même définition mais en remplaçant le print("coucou") par le mot clé return et 'coucou'
def ma_fonction():
  return "coucou"

# afficher l'appel de la fonction
print("appel de ma_fonction() => ", ma_fonction())
# alternativement, on peut affecter l'appel de la fonction dans une variable
ret = ma_fonction()
```

```python
# -- définition d'une fonction avec des paramètres et appel des paramètres ---

# définir une fonction qui ajoute 2 entiers en paramètres et qui retourne la somme 
# a, b dans la définition de la focntions sont des paramètres formels

def addition(a, b):
    return a + b

# injection des variables globales en tant que paramètres d'appel: passage en référence
print(f"appel de addition(2.5, 3.14) => {addition(2.5, 3.14)}")
# attention au typage dynamique !!!
print(f"appel de addition('ff', 'rr') => {addition('ff', 'rr')}")
```

```python
# ------------ annotations, documentation et contrôle -------------------


# créer une fonction division de 2 floats qui retourne un float
# les annotations ne sont pas des contrôles, ce ne sont que des indications

def division(a: float, b: float) -> float | None:
    """Divise a par b et retourne le résultat.

    Args:
        a (float): le numérateur
        b (float): le dénominateur

    Returns:
        float: le résultat de la division

    Raises:
        ValueError: si b est égal à 0
    """
    if b != 0:
        return a / b

# 1. en notifiant les types d'entrée et de sortie de la fonction
# 2. documenter la fonction en ajoutant une """doctstring""" au début du bloc
# 3. faire un contrôle sur le dénominateur
```

```python
# --------------- types de paramètres ----------------------------------

# créer une fonction calcul_tva qui prend 2 paramètres le prix ht et le taux
# et retourne la valeur de tva sur ce prix et ce taux
# 1. annotations et documentation
# 2. le prix est arbitraire => paramètre positionnel / obligatoire
# MAIS on considère que la valeur par défaut du taux est 20 => paramètre nommé / optionnel

def calcul_tva(prix_ht: float, taux: float = 20) -> float:
    """Calcule la TVA sur un prix HT donné et un taux donné.

    Args:
        prix_ht (float): le prix hors taxes
        taux (float, optional): le taux de TVA. Defaults to 20.

    Returns:
        float: la valeur de la TVA
    """
    return prix_ht * taux / 100
# appeler la fonction sans paramètres

print(f"""appel positionnel: 199 -> prix_ht, 5.5 -> taux 
       => {calcul_tva(199, 5.5)}""")
print(f"""appel avec un paramètre optionnel: 199 -> prix ht et rien -> taux
       => {calcul_tva(199)}""")
print(f"""appel nommé: les valeurs sont fléchées vers les paramètres 
       => pas besoin d'ordre => {calcul_tva(prix_ht=199, taux=5.5)}""")
```

```python
# --------------------- passage en référence des paramètres ------------------

def func(param):
    print(f"param: {param}, id(param): {id(param)}")
    return param + 2

glob = 10
print(f"glob: {glob}, id(glob): {id(glob)}")

glob = func(glob)
print(f"glob après reafectation: {glob}, id(glob): {id(glob)}")

## l'id précédent (glob/ param) n'a plus de référence ===> Garbage Collector
```

```python
# ------------- passage en référence d'un type muable -----------
from typing import Any

def my_append(l: list, val: Any) -> None:
    print(f"param: {l}, id(param): {id(l)}")
    l.append(val)

glob = [1, 2, 3]
print(f"glob: {glob}, id(glob): {id(glob)}")

my_append(glob, 4)
print(f"glob après injection dans la fonction: {glob}, id(glob): {id(glob)}")
```

```python
# --------------------- portée globale /locale ------------------

def func(param):
    x = 2 
    print(f"x local: {x}, id(x): {id(x)}")
    return param + x

x = 10
print(f"x: {x}, id(glob): {id(x)}")

x = func(x)
print(f"x après reafectation: {x}, id(x): {id(x)}")
```

```python
# -------------- modifier des globales dans une PROCEDURE ------------
def func():
    global x
    x += 2
    

x = 10
print(f"x: {x}, id(glob): {id(x)}")

func()
print(f"x après exécutionde proc: {x}, id(x): {id(x)}")
```

```python
# ------------- programmation fonctionnelle en python ---------------
from typing import Callable
## pricipe:  le programme principal ressemble à Y = F°G°H°...°Z(X)

data = [ x for x in range(10)]

def square(x: float) -> float:
    return x ** 2

def my_map(f: Callable, _input: list):
    for i in range(len(_input)):
        _input[i] = f(_input[i])
    return _input

my_map(square, data)


```

```python
# ------------------ map & fonction lambda --------------------------
data = [ x for x in range(10)]

# def square(x: float) -> float:
#     return x ** 2

# l'itération du map est exécutée en C natif ==> plus rapide que for !!
list(map(lambda x: x ** 2, data))

```

```python
# ------------------ filter --------------------------
data = [ x for x in range(10)]

# je veux filtrer les données d'entrée pour préserver les impairs !!
# les élements "x" de data qui retourne True à la lambda seront en sortie
list(filter(lambda x: x % 2, data))

# mauvaise pratique !!!
# (lambda x: x%2)(1)
```
