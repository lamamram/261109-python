```python
# --------------------- décorateurs ---------------------

# décorateur = une fonction 
# qui utilise une fonction en param. 
# et qui retourne une fonction (wrapper) 
# qui elle même étends,  complète la fct en param.

def strong(f):
    def wrapper(*args, **kwargs):
        return f"<strong>{f(*args, **kwargs)}</strong>"
    return wrapper

@strong
def to_upper(s: str):
    return s.upper()

def to_lower(s: str):
    return s.lower()

# to_upper = strong(to_upper)

to_upper("bonjour")
```

```python
# --------------------- décorateurs ---------------------

def strong(f):
    def wrapper(*args, **kwargs):
        # f.__annotations__, f.__doc__
        return f"<strong>{f(*args, **kwargs)}</strong>"
    return wrapper

def em(f):
    def wrapper(*args, **kwargs):
        return f"<em>{f(*args, **kwargs)}</em>"
    return wrapper

# to_upper = strong(to_upper)
@strong
@em
def to_upper(s: str):
    return s.upper()

def to_lower(s: str):
    return s.lower()


to_upper("bonjour")
```

```python
# ----------------- décorateur paramétré ----------------
def tag(tag_name):
    """fermeture en python"""
    def decor(f):
        """décorateur"""
        def wrapper(*args, **kwargs):
            return f"<{tag_name}>{f(*args, **kwargs)}</{tag_name}>"
        return wrapper
    return decor

@tag("strong")
@tag("em")
def to_upper(s: str):
    return s.upper()

to_upper("bonjour")
```
